```python
import numpy as np
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from neural_regression import NeuralNetwork
```

```python
train=pd.read_csv("train_nn.csv")
test=pd.read_csv("test_nn.csv")
```

```python
X = train.iloc[:,1:-2].values
y_cls = train.iloc[:, -2].values
y_reg = train.iloc[:, -1].values
y_cls=y_cls.reshape(-1,1)
y_reg=y_reg.reshape(-1,1)
X_test=test.iloc[:,1:].values
```

```python
np.random.seed(42)

n = X.shape[0]
indices = np.random.permutation(n)

split = int(0.8 * n)
train_idx = indices[:split]
val_idx   = indices[split:]

X_train, X_val = X[train_idx], X[val_idx]
y_train_reg, y_val_reg = y_reg[train_idx], y_reg[val_idx]



```

```python
mean = X_train.mean(axis=0)
std = X_train.std(axis=0)
X_train=(X_train-mean)/std
X_val=(X_val-mean)/std
X_test=(X_test-mean)/std

```

```python
model = NeuralNetwork(
    layer_sizes=[20, 32, 16, 1],
    lr=0.01
)

model.fit(X_train, y_train_reg, epochs=1000)
preds = model.predict(X_val)





print(preds[:10])
```

```python
def r2_score(y, y_pred):
    y = y.reshape(-1)
    y_pred = y_pred.reshape(-1)
    ss_res = np.sum((y - y_pred)**2)
    ss_tot = np.sum((y - np.mean(y))**2)
    return 1 - (ss_res / ss_tot)
```

```python
print("Val   R2:", r2_score(y_val_reg, preds))
```

```python
test_preds=model.predict(X_test)
```

```python
neural_regression = pd.DataFrame({
    "target": test_preds.ravel()
})

neural_regression.to_csv("neural_regression.csv", index=False)
print("neural_regression.csv created!")

```
