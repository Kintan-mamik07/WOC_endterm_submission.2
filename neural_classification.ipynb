```python
import numpy as np
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from neural_classification import NeuralNetwork




```

```python
train=pd.read_csv("train_nn.csv")
test=pd.read_csv("test_nn.csv")
```

```python
X = train.iloc[:,1:-2].values
y_cls = train.iloc[:, -2].values
y_reg = train.iloc[:, -1].values
y_cls=y_cls.reshape(-1,1)
y_reg=y_reg.reshape(-1,1)
X_test=test.iloc[:,1:].values
```

```python
np.random.seed(42)

n = X.shape[0]
indices = np.random.permutation(n)

split = int(0.8 * n)
train_idx = indices[:split]
val_idx   = indices[split:]

X_train, X_val = X[train_idx], X[val_idx]
y_train_cls, y_val_cls = y_cls[train_idx], y_cls[val_idx]


```

```python
mean = X_train.mean(axis=0)
std = X_train.std(axis=0)
X_train=(X_train-mean)/std
X_val=(X_val-mean)/std
X_test=(X_test-mean)/std
```

```python
model = NeuralNetwork(layer_sizes=[20, 64,32 ,27 ], lr=0.01)

model.fit(X_train, y_train_cls, epochs=300)

preds = model.predict(X_val)
print(preds[:10])
```

```python
from sklearn.metrics import f1_score

macro_f1 = f1_score(y_val_cls,preds, average="macro")
print(macro_f1)
```

```python
test_preds = model.predict(X_test)
neural_classification = pd.DataFrame({
    "target": test_preds
})

neural_classification.to_csv("neural_classification.csv", index=False)
print("neural_classification.csv created!")



```
