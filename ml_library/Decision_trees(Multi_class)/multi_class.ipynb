```python
import pandas as pd
import numpy as np
from random_forest import DecisionTree
```

```python
df=pd.read_csv("train_multi_class.csv")
X=df.iloc[:,:-1].values
y=df.iloc[:,-1].values
```

```python
mask = ~pd.isna(y)
X = X[mask]
y = y[mask]
```

```python
np.random.seed(42)

n = X.shape[0]
indices = np.random.permutation(n)

split = int(0.8 * n)
train_idx = indices[:split]
cv_idx   = indices[split:]

X_train, X_cv = X[train_idx], X[cv_idx]
y_train, y_cv = y[train_idx], y[cv_idx]
```

```python
mean=X_train.mean(axis=0)
std=X_train.std(axis=0)
X_train=(X_train-mean)/std
X_cv=(X_cv-mean)/std
```

```python
model = DecisionTree(
    max_depth=10,
    min_samples=50,
    
)

model.fit(X_train, y_train)
```

```python
df=pd.read_csv("test_multi_class.csv")
X_test=df.values
X_test=(X_test-mean)/std
```

```python
preds = model.predict(X_cv)
```

```python
print(preds[:10])
accuracy = np.mean(preds == y_cv)
print("Cross validation Accuracy:", accuracy)

```

```python
test_preds=model.predict(X_test)
```

```python
logistic = pd.DataFrame({
    
    "label": test_preds
})

logistic.to_csv("multi_class.csv", index=False)
print("multi_class.csv created!")

```
